# Stage 5: cited retrieval
Run `python -m src.rag --download-model` once after Stage 4; later runs use `python -m src.rag` and cached local weights. The five knowledge files are explicitly nonclinical placeholders. Replace them using data/knowledge/README.md before claiming medical grounding. The tested module uses sentence-transformers, FAISS and a no-API cited template. See README.md for the reusable GroundedAssistant API and optional validated HTTP adapter. Reports are in reports/rag_report.md and the readable example is in data/processed/stage5/example_explanation.md. The Stage 6 interface is implemented in app/app.py; launch with python -m streamlit run app/app.py after setup.
