# Stages 2 and 3: features and modeling
Use `python -m src.features` and `python -m src.train` from the repository root after installing requirements and running tests. The tested source modules implement fold-specific preprocessing, patient-grouped holdout/CV, three models and model reports. See README.md for commands, artifacts and retrospective limitations. Results are in reports/model_comparison.md; explainability is reserved for Stage 4.
