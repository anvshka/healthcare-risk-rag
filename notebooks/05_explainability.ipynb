# Stage 4: SHAP explainability
Run `python -m src.explain` from the repository root after Stage 3. The tested module produces global and individual probability-space explanations without refitting the model. See README.md for `RiskExplainer.explain_admission(hadm_id)` and reports/explainability_report.md for methods and limitations. Positive/negative attributions explain model behavior and do not establish medical causation. Literature-grounded explanations remain Stage 5.
